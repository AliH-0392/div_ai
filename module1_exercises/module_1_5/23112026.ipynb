```python
class Customer:

    def __init__(self, name, age):

        self.name = name
        self.age = age

    def ageGroup(self):

        if(self.age <= 25):
            return "Young"
        elif(self.age > 25):
            return "Middleaged"
        else:
            return "Elderly"

fayl = open(pth, encoding="utf-8")

encoder_s = csv.DictReader(fayl)
reader_s = csv.DictReader(fayl)

for line in reader_s:

    c = Customer(line["ad"], int(line["yas"]))
    print(c.name, c.age, c.ageGroup())

fayl.close()
```
