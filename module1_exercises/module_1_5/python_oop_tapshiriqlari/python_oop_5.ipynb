```python
class User:

    def __init__(self, username, email, password):
        self.username = username
        self._email = email
        self.__password = password

    def check_password(self, user_password):
        return self.__password == user_password
```

```python
user_1 = User("Admin", "root_admin@mail.com", "adminroot")
user_pw = input("Enter your pass: ")

if (user_1.check_password(user_pw)):
    print("correct")
else:
    print("incorrect")
```
