```python
class BankAccount:

    def __init__(self, name, balance):
        
        self.name = name
        self.balance = balance


class SavingsAccount(BankAccount):
    
    def __init__(self, name, balance):
        
        super().__init__(name, balance)
        #self.interest = interest

    def calculate_interest(self):
        #interest = self.balance * 1.05
        return self.balance * 1.05
        #return interest
```

```python
sav_acc_1 = SavingsAccount("ali", 500)

print(sav_acc_1.calculate_interest())
```

```python
class CreditCard:

    def __init__(self, amount, balance):

        self.amount = amount
        self.balance = balance
        ...
    
    def pay(self, amount):

        return self.balance - (self.amount * 1.02) 



class PayPal:

    def __init__(self, amount, balance):

        self.amount = amount
        self.balance = balance
    
    def pay(self, amount):

        return self.balance - (self.amount * 1.03)
```

```python
payment_1 = CreditCard(500, 5000)
payment_2 = PayPal(500, 5000)

print(f"Balance: {payment_1.balance}, amount: {payment_1.amount}")
print(f"after paying with a credit card {payment_1.pay(500)}")

print("\n")

print(f"Balance: {payment_2.balance}, amount: {payment_2.amount}")
print(f"After paying with PayPal: {payment_2.pay(500)}")

```

```python
payment_list = [payment_1, payment_2]

for payment in payment_list:

    print(f"After paying with {payment.__class__.__name__} the balance will be: {payment.pay(500)}")
    #print(payment.pay(500))
```
