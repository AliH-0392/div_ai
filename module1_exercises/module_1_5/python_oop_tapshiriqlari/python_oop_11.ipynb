```python
class Person:

    def __init__(self, name, age):

        self.name = name
        self.age = age
```

```python
class Employee(Person):

    def __init__(self, name, age, salary):

        super().__init__(name, age)
        self.salary = salary

    def __str__(self):
        return f"Name: {self.name}. Age: {self.age}, salary: {self.salary}"
```

```python
employee_1 = Employee("Ali", 25, 3000)

print(employee_1)
```
