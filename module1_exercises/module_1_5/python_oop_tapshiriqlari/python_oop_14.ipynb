```python
class Employee:

    def __init__(self, name, position, salary):

        self.name = name
        self.position = position
        self.salary = salary

    def calculateSalary(self):

        return self.salary

    def __repr__(self):

        return f"Name: {self.name}. Position: {self.position}. Salary: {self.salary}"
```

```python
class Manager(Employee):

    def __init__(self, name, position, salary, bonus):

        super().__init__(name, position, salary)
        self.bonus = bonus

    def calculateSalary(self):

        return self.salary + self.bonus

    def __repr__(self):

        return f"Name: {self.name}. Position: {self.position}. Salary: {self.salary + self.bonus} ({self.bonus} bonus)"
```

```python
manager_1 = Manager("Elvin", "CEO", 40000, 8000)
employee_1 = Employee("Ali", "Technician", 4000)
```

```python
print(manager_1.calculateSalary())
print(employee_1.calculateSalary())
```
