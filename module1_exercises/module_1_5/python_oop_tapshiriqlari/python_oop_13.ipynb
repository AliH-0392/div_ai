```python
class Book:

    def __init__(self, name, author,release_date):

        self.name = name
        self.author = author
        self.release_date = release_date

    def __repr__(self):

        return f"Book name: {self.name}\nAuthor: {self.author}\nRelease date: {self.release_date}"

    def __eq__(self, other):

        return self.name == other.name and self.author == other.author and self.release_date == other.release_date
```

```python
book_1 = Book("Bleed", "Robin Cook", 2026)
book_2 = Book("Cold Redemption", "Nalini Singh", 2026)
book_3 = Book("Bleed", "Robin Cook", 2026)

print(book_1 == book_2)
print(book_1 == book_3)
print("\n")
print(book_3)
```
