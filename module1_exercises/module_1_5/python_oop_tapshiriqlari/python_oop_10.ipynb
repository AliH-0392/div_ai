```python
class Student:

    uni_name = "ADNSU"

    def __init__(self, stu_name):

        self.stu_name = stu_name

    @classmethod
    def change_uni(cls, new_uni):

        cls.uni_name = new_uni

    def __str__(self):

        return f"Student {self.stu_name}, Uni: {self.uni_name}"
```

```python
student_1 = Student("Ali")

print(student_1)

Student.change_uni("AzTU")

print(student_1)
```
