```python
class Product: 

    def __init__(self, name, price, category):

        self.name = name
        self.price = price
        self.category = category
    
    def __str__(self):
        return f"The product {self.name} belongs to {self.category} category and costs {self.price} AZN"
```

```python
product_1 = Product("Asus Zenbook 14", 3000,"Compsuters")
product_2 = Product("Oppo Find x9 Ultra", 4000, "Smartphones")
#product_3 = 

print(product_1)
print(product_2)
```
