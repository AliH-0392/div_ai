```python
class Student:

    schoolName = "ADNSU"

    def __init__(self, name, score):

        self.name = name
        self.score = score

    def describe(self):
        
        return f"Name: {self.name}, grade: {self.score}, University: {self.schoolName}"
```

```python
student_1 = Student("Ali", 84)
student_2 = Student("Nihat", 81)

student_1.describe()

#student_1.schoolName
```
