```python
class Car:
    
    wheels = 4

    def __init__(self, make, model, year):
        self.make = make #Ford
        self.model = model #mustang
        self.year = year

    def Describe(self):
        return f"This is a {self.year} {self.make} {self.model}"


myCar = Car("Mercedes-Benz", "560SEL", 1989)

print(myCar.Describe())
```

```python
class ElectricCar(Car):

    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year)
        #super().__init__(model)
        #super().__init__(year)
        self.battery_size = battery_size

    def Describe(self):
        return f"{super().Describe()} with a {self.battery_size} kWh battery"

    def refuel(self):
        return f"The car is charging."

myEC = ElectricCar("Tesla", "Model S", 2026, 100)

print(myEC.Describe())
```

```python
class FuelCar(Car):

    def __init__(self, make, model, year, fuel_type):
        super().__init__(make, model, year)
        self.fuel_type = fuel_type

    def refuel(self):
        return f"The car is fueling."

myFC = FuelCar("Mercedes-Benz", "E220 W124", 1995, "Petrol/Gasoline")
```

```python
cars = [myEC, myFC]

for car in cars:
    print(car.refuel())
```

```python
class BankAccount:

    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance

    #@property
    def deposit(self, add):
        self.__balance += add
    
    #@property
    def withdraw(self, subtract):
        self.__balance -= subtract

    @property
    def get_balance(self):
        return self.__balance 
```

```python
myAcc = BankAccount("Ali", 5000)

myAcc.deposit(3000)
myAcc.get_balance

myAcc.withdraw(2000)
myAcc.get_balance
```

```python
from math import sqrt
import math

class Line:
    def __init__(self, point1, point2):
        self.point1 = point1
        self.point2 = point2

    def calculate_length(self):
        x1, y1 = self.point1
        x2, y2 = self.point2
        len = sqrt((x2-x1) ** 2 + (y2-y1) ** 2)
        return len

    def calculate_slope(self):
        x1, y1 = self.point1
        x2, y2 = self.point2
        slope = (y2-y1) / (x2-x1)
        return slope
```

```python
p1 = (3, 2)
p2 = (8, 10)
line1 = Line(p1, p2)

print(f"line1 length: {line1.calculate_length()}")
print(f"line1 slope: {line1.calculate_slope()}")
```

```python
class Rectangle:

    def __init__(self, width, height):
        self.height = height
        self.width = width
    
    def calculate_area(self):
        return self.height * self.width

    def calculate_perimeter(self):
        return 2 * (self.width + self.height)
```

```python
rectangle1 = Rectangle(2, 5)

print(f"perimeter: {rectangle1.calculate_perimeter()}")
print(f"area: {rectangle1.calculate_area()}")
```

```python
class Cylinder:

    def __init__(self, radius, height):
        self.radius = radius
        self. height = height

    def calculate_volume(self):
        return pow(self.radius, 2) * self.height * math.pi
    
    def calculate_surface_area(self):
        return 2 * math.pi * self.radius * (self.radius + self.height)
```

```python
cylinder1 = Cylinder(4, 7)

print(f"Surface area: {cylinder1.calculate_surface_area()}")
print(f"Volume: {cylinder1.calculate_volume()}")
```

```python
"""Create a Book class with an __init__ method that takes title, author, and pages.
Implement the __str__ special method to return a user-friendly string representation, e.g., "'The Hobbit' by J.R.R. Tolkien, 310 pages".
Implement the __len__ special method to return the number of pages in the book.
Create a Book object and test both print(my_book) and len(my_book)."""

class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    def __str__(self):
        return f"{self.title} by {self.author}, {self.pages} pages"
    
    def __len__(self):
        return self.pages
```

```python
myBook = Book("The Hobbit", "J.R.R Tolkien", 310)

print(myBook)
print(len(myBook))
```
