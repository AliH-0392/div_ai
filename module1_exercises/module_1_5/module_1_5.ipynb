```python
class Car:
    
    wheels = 4

    def __init__(self, make, model, year):
        self.make = make #Ford
        self.model = model #mustang
        self.year = year

    def Describe(self):
        return f"This is a {self.year} {self.make} {self.model}"


myCar = Car("Mercedes-Benz", "560SEL", 1989)

print(myCar.Describe())
```

```python
class ElectricCar(Car):

    def __init__(self, make, model, year, battery_size):
        super().__init__(make, model, year)
        #super().__init__(model)
        #super().__init__(year)
        self.battery_size = battery_size

    def Describe(self):
        return f"{super().Describe()} with a {self.battery_size} kWh battery"

myEC = ElectricCar("Tesla", "Model S", 2026, 100)

print(myEC.Describe())
```
