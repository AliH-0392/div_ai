```python
accounts = {
    "Elton": {"password": "1234", "balance": 1500},
    "Ali": {"password": "5678", "balance": 800},
    "Aysel": {"password": "9999", "balance": 2500}
}
```

```python
username = input("İstifadəçi adı daxil et: ")


if username in accounts:
    
    password = input("Kodu daxil edin: ")

    if password == accounts[username]["password"]:
        print(f"Xoş gəldiniz, {username}!")

        balans = accounts[username]["balance"]

        print(f"Cari balansınız: {balans} AZN")
    else:
        print("Şifrə yanlışdır")

```

```python
tesdiq = bool(int(input("Nağdlaşdırma etmək istəyirsiniz? (0 və ya 1)")))

if(tesdiq == True):

    daxil_edilen_mebleg = int(input("Meblegi daxil edin: "))

    if(daxil_edilen_mebleg <= 0):
        print("Yanlış məbləğ!")
    elif(daxil_edilen_mebleg > balans):
        print("Balansınızda kifayət qədər vəsait yoxdur!")
    elif(daxil_edilen_mebleg < balans):
        balans -= daxil_edilen_mebleg
        print(f"Balansınızdan {daxil_edilen_mebleg} qədər məbləğ çıxıldı.\nYeni balansınız: {balans}")
    
else:
    print("Əməliyyatdan imtina edildi.")
```
