```python
sheherler = {
    "Bakı": [14, 15, 16, 14, 17, 15, 16],
    "Gəncə": [12, 11, 13, 14, 12, 11, 13],
    "Lənkəran": [16, 17, 18, 16, 15, 17, 18],
    "Şəki": [8, 9, 7, 10, 8, 9, 7],
    "Quba": [5, 6, 4, 5, 7, 6, 5]
}

heftenin_gunleri = ["Bazar ertəsi", "Çərşənbə axşamı", "Çərşənbə", "Cümə axşamı", "Cümə", "Şənbə", "Bazar"]
```

```python
for sheher in sheherler.keys():
    print(sheher)
```

```python
for sheher, temperatur in sheherler.items():
    orta_temp = sum(temperatur) / len(temperatur)
    print(f"{sheher} üçün həftəlik orta temperatur: {orta_temp:.2f}C")
```

```python
for sheher, temperatur in sheherler.items():
    for index in range(len(temperatur)):
        hefte_gunu = heftenin_gunleri[index]
    print(f"Həftənin ən soyuq günü {sheher} şəhəri üçün: {min(temperatur)} dərəcə")
    print(f"Həftənin ən isti günü {sheher} şəhəri üçün: {max(temperatur)} dərəcə")
    
```
