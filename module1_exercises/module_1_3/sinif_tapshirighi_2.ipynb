```python
def faktorial(eded):

    qaytar = 1

    for i in range(2, eded + 1):
        qaytar *= i

    return qaytar
```

```python
eded = int(input("Ededi daxil edin: "))
print(f"{eded} ededinin faktoriali: {faktorial(eded)}")
#faktorial(5)
```

```python
def ededin_ededleri_cemi(eded):

    cem = 0

    while eded > 0:
        
        reqem = eded % 10
        cem += reqem
        eded //= 10

    return cem
```

```python
test_eded = 456

print(test_eded//10)
print(test_eded%10)
```

```python
eded_1 = int(input("Ededi daxil edin: "))
print(f"{eded_1} ededinin ededleri cemi: {ededin_ededleri_cemi(eded_1)}")
```

```python
import random

secimler = ["daş", "kağız", "qayçı"]

for i in range(1, 4):
    print(f"raund {i}: ")

    istifadeci = input("Seciminizi edin (daş, kağız, qayçı)")

    if istifadeci not in secimler:
        print("Düzgün seçim etmədiniz")
        continue
        
    komputer = random.choice(secimler)

print(f"Sizin seçiminiz: {istifadeci}")
print(f"Kompüterin seçimi: {komputer}")

if komputer == istifadeci:
    print("Heç-heçə")
elif (istifadeci == "daş" and komputer == "qayçı") or \
     (istifadeci == "kağız" and komputer == "daş") or \
     (istifadeci == "qayçı" and komputer == "kağız"):
       print("Siz qalıb gəldiniz!")
    
```
