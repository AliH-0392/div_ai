```python
mashinlar = {
    "Honda" : 20000,
    "Kia" : 40000,
    "BMW" : 30000,
    "Lada" : 15000
}

mashinlar
```

```python
bosh_list = []

for mashin_adi, mashin_qiymeti in mashinlar.items():
    if mashin_qiymeti > 20000:
        bosh_list.append("Bahali")
        #print("Bahali")
    else:
        bosh_list.append("Ucuz")
        #print("Ucuz")

bosh_list
```

```python
def mashin_qiymeti_qaytar(t):
    mashin_adi, mashin_qiymeti = t
    if(mashin_qiymeti > 20000):
        return "Bahali"
    else:
        return "Ucuz"
    return


list(map(mashin_qiymeti_qaytar, mashinlar.items()))
```
