```python
ilk_eded = 13
eded = ilk_eded
ikili_say_string = ""

while eded > 0:

    #netice = eded // 2
    qaliq = eded % 2
    eded //= 2
    ikili_say_string += str(qaliq)


ikili_say_eded = (int(ikili_say_string[::-1]))

print(f"{ilk_eded} ededinin ikili say sistemi yazilishi: {ikili_say_eded}")

birleri_say = 0
sifirlari_say = 0

for i in range(len(ikili_say_string)):
    #print(ikili_say_string[i])
    if int(ikili_say_string[i]) == 1:
        birleri_say += 1
    elif int(ikili_say_string[i]) == 0:
        sifirlari_say += 1

print(f"Birlerin sayi: {birleri_say}")
print(f"Sifirlarin sayi: {sifirlari_say}")
```

```python
matris = [
    [1, 2, 3], 
    [4, 5, 6], 
    [7, 8, 9]
]

esas_diaqonal_cem = 0
komekchi_diaqonal_cem = 0
esas_diaqonal_elementleri = []
komekchi_diaqonal_elementleri = []

#for i in matris:
 #   for j in matris:
  #      if(i == j):
   #         esas_diaqonal_cem += matris[i][j]

n = len(matris)

for i in range(n):
    esas_diaqonal_elementi = matris[i][i]
    esas_diaqonal_cem += esas_diaqonal_elementi
    esas_diaqonal_elementleri.append(esas_diaqonal_elementi)

    komekchi_diaqonal_elementi = matris[i][n - 1 - i]
    komekchi_diaqonal_cem += komekchi_diaqonal_elementi
    komekchi_diaqonal_elementleri.append(komekchi_diaqonal_elementi)



print(f"Esas diaqonal cem: {esas_diaqonal_cem}")
print(f"Komekci diaqonal cem: {komekchi_diaqonal_cem}")
print(f"esas diaqonalin maksimumu: {max(esas_diaqonal_elementleri)}")
if esas_diaqonal_cem == komekchi_diaqonal_cem:
    print("Cemler beraberdir")
else:
    print("cemler beraber deyil")
```

```python
# ad ve yash 
# insanlar = { ad : yash }

istifadeciler = [
    {"ad": "Əli", "yash": 17},
    {"ad": "Aygün", "yash": 24},
    {"ad": "Nərmin", "yash": 35},
    {"ad": "Rüstəm", "yash": 45}
]

yash_qrupu = {"minor" : [], "young": [], "adult": [], "senior" : []}

for adam in istifadeciler:
   

    if(adam["yash"] >= 0):
        if(adam["yash"] < 18):
            yash_qrupu["minor"].append(adam["ad"])
        elif((adam["yash"] >= 18) and (adam["yash"] <= 30)):
            yash_qrupu["young"].append(adam["ad"])
        elif((adam["yash"] >= 31) and (adam["yash"]) <= 40):
            yash_qrupu["adult"].append(adam["ad"])
        elif(adam["yash"] > 40):
            yash_qrupu["senior"].append(adam["ad"])


for yash_qrup_adiqrup_adi, uzvler in yash_qrupu.items():
    print(f"{yash_qrup_adiqrup_adi}: {uzvler}")
```
